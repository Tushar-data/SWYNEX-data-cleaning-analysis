### Matplotlib liabaries

```python
import matplotlib.pyplot as plt
```

```python
# churn distribution
plt.figure(figsize=(6,4))
churn_counts =df["Churn"].value_counts()

plt.bar(churn_counts.index, churn_counts.values)
plt.xlabel("Churn")
plt.ylabel("Number of Customers")
plt.title("Customer Chrun Distribution")
plt.show()
```

```python
# contract distribution
plt.figure(figsize=(7,4))
contract_counts=df["Contract"].value_counts()

plt.bar(contract_counts.index, contract_counts.values)
plt.xlabel("Contract Type")
plt.ylabel("Number of Customer")
plt.title("Customer Distribution of Contract")
plt.xticks(rotation=20)
plt.show()
```

```python
# senior citizen distribution
senior_count = df["Senior_Citizen"].value_counts()
plt.figure(figsize=(6,4))
plt.bar(senior_count.index, senior_count.values)
plt.title("Senior Citizen Distribution")
plt.xlabel("Senior Citizen")
plt.ylabel("Number of Customers")
plt.show()
```

```python
# gender distribution
gender_count = df["Gender"].value_counts()
plt.figure(figsize=(6,4))
plt.bar(gender_count.index, gender_count.values)
plt.title("Customer Distribution by Gender")
plt.xlabel("Gender")
plt.ylabel("Number of Customers")
plt.show()
```

```python
# partner distribution 
partner_count=df["Partner"].value_counts()
plt.figure(figsize=(6,4))
plt.bar(partner_count.index, partner_count.values)
plt.title("Customer Distrubution by Partner")
plt.xlabel("Partner")
plt.ylabel("Customers")
plt.show()
```

```python
# dependent distrubution 
dependent_count=df["Dependents"].value_counts()
plt.figure(figsize=(6,4))
plt.bar(dependent_count.index, dependent_count.values)
plt.title("Customers Distribution by Dependents")
plt.xlabel("Dependents")
plt.ylabel("Customers")
plt.show()
    
```

```python
# monthly charges distribution
plt.figure(figsize=(7,4))
plt.hist(df["Monthly_Charges"],bins=20)
plt.title("Monthly Charges Distribution")
plt.xlabel("Number of Customers")
plt.show()
```

```python
# tenure distribution
plt.figure(figsize=(7,4))
plt.hist(df["Tenure_Months"], bins=15)
plt.title("Customers Tenure Distribution")
plt.xlabel("Monthly Charges")
plt.ylabel("Number of Customers")
plt.show()
```

```python
# monthly charges vs tenure
plt.figure(figsize=(7,5))
plt.scatter(df["Tenure_Months"], df["Monthly_Charges"])
plt.title("Tenure vs Monthly Charges")
plt.xlabel("Tenure Months")
plt.ylabel("Monthly Charges")
plt.show()
```
